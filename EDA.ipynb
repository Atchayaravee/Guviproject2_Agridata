```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Atchaya Raveendran\Desktop\agri data project2\ICRISAT-District Level Data - ICRISAT-District Level Data.csv")
df
```

```python
df.columns = (
    df.columns.str.strip()  # remove leading/trailing spaces
              .str.replace(' ', '_')
              .str.replace('(', '', regex=False)
              .str.replace(')', '', regex=False)
              .str.replace('-', '_')
              .str.replace('/', '_')
              .str.replace('__', '_')
              
              .str.lower()
)


```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# Count duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python

# Check for negative values in numeric columns
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns

negatives = (df[numeric_cols] < 0).sum()
print("Negative values in columns:\n", negatives[negatives > 0])
```

```python
df[numeric_cols] = df[numeric_cols].applymap(lambda x: x if x >= 0 else 0)
negatives_after = (df[numeric_cols] < 0).sum()
print("Still negative values:\n", negatives_after[negatives_after > 0])
```

```python
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
template_style = "plotly_dark"

# 1. Top 7 Rice Producing States (Bar plot)
rice_top7 = df.groupby("state_name")["rice_production_1000_tons"].sum().sort_values(ascending=False).head(7).reset_index()
print(rice_top7)
fig1 = px.bar(rice_top7, x="state_name", y="rice_production_1000_tons",
              title="Top 7 Rice Producing States",
              text="rice_production_1000_tons",
              template=template_style,
              color="state_name")
fig1.update_traces(textposition='inside', textfont_size=10, showlegend=False)
fig1.update_layout(yaxis_title="Production (1000 tons)", xaxis_title="State")
fig1.show()
```

```python
area_prod = df.groupby("year")[["rice_area_1000_ha", "rice_production_1000_tons",
                                 "wheat_area_1000_ha", "wheat_production_1000_tons",
                                 "maize_area_1000_ha", "maize_production_1000_tons"]].sum().reset_index()
fig15 = px.scatter(area_prod, x="rice_area_1000_ha", y="rice_production_1000_tons",
                   size="wheat_area_1000_ha", color="maize_production_1000_tons",
                   title="Impact of Area on Production (Rice, Wheat, Maize)",
                   template=template_style,
                   hover_data=["wheat_production_1000_tons", "maize_production_1000_tons"],
                   size_max=20,
                   color_continuous_scale=px.colors.sequential.Plasma)
fig15.update_layout(xaxis_title="Area (1000 ha)", yaxis_title="Production (1000 tons)")
fig15.show()
```

```python
area_prod = df.groupby("year")[[
    "rice_area_1000_ha", "rice_production_1000_tons",
    "wheat_area_1000_ha", "wheat_production_1000_tons",
    "maize_area_1000_ha", "maize_production_1000_tons"
]].sum().reset_index()

# Prepare a list to collect data for each crop
data_list = []

for crop in ['rice', 'wheat', 'maize']:
    temp = area_prod[['year', f'{crop}_area_1000_ha', f'{crop}_production_1000_tons']].copy()
    temp.columns = ['year', 'area', 'production']  # rename for consistency
    temp['crop'] = crop
    data_list.append(temp)

# Combine all crops into one DataFrame
plot_df = pd.concat(data_list, ignore_index=True)

# Plot
fig = px.scatter(
    plot_df,
    x='area',
    y='production',
    color='crop',
    hover_data=['year'],
    title="Impact of Area on Production (Rice, Wheat, Maize)",
    labels={'area': 'Area (1000 ha)', 'production': 'Production (1000 tons)', 'crop': 'Crop'},
    template=template_style
)
fig.show()
```

```python
yield_states = df.groupby("state_name")[["rice_yield_kg_per_ha", "wheat_yield_kg_per_ha"]].mean().reset_index()
# Convert wide data to long format
yield_states_long = yield_states.melt(
    id_vars='state_name',
    value_vars=['rice_yield_kg_per_ha', 'wheat_yield_kg_per_ha'],
    var_name='Crop',
    value_name='Yield'
)

# Rename crops for clearer labels
yield_states_long['Crop'] = yield_states_long['Crop'].replace({
    'rice_yield_kg_per_ha': 'Rice Yield (kg/ha)',
    'wheat_yield_kg_per_ha': 'Wheat Yield (kg/ha)'
})

# Create grouped bar chart
fig16 = px.bar(
    yield_states_long,
    x='state_name',
    y='Yield',
    color='Crop',
    barmode='group',
    title="Rice vs Wheat Yield Across States",
    template=template_style
)

# Adjust axis labels and tick angle
fig16.update_layout(
    xaxis_title="State",
    yaxis_title="Yield (kg/ha)",
    xaxis_tickangle=-45
)

fig16.show()
```
